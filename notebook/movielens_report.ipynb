# MovieAnalysis

## Описание проекта
Проект представляет собой автоматизированный ETL-пайплайн для работы с кинематографическими данными. Основной фокус направлен на слияние большого массива зрительских рейтингов с финансовой аналитикой, извлеченной с платформы TMDB посредством веб-скрейпинга.

В этом отчете представлены итоговые аналитические витрины, готовые для дальнейшего исследования и бизнес-анализа.

```python
import sqlite3
import pandas as pd

DB_PATH = '../dwh.sqlite'

def query_sql(query):
    """A helper function for quickly reading marts in a pandas DataFrame"""
    with sqlite3.connect(DB_PATH) as conn:
        return pd.read_sql(query, conn)
```

### Самые продуктивные режиссеры индустрии

В этой витрине мы оцениваем режиссеров не по кассовым сборам или оценкам критиков, а по их **кинематографической плодовитости** — количеству картин, представленных в нашем датасете.

```python
query_directors = "SELECT * FROM mart_top_directors LIMIT 10"
df_directors = query_sql(query_directors)

df_directors
```

### Финансовые амбиции и риски: Самые дорогие фильмы

Бюджет картины — это прямой показатель ставки, которую делает студия, и уровня принимаемого финансового риска. В этой витрине мы объединяем базовый справочник фильмов с распарсенной статистикой TMDB, чтобы выявить настоящие мега-блокбастеры.

```python
query_expensive = "SELECT * FROM mart_most_expensive LIMIT 10"
df_expensive = query_sql(query_expensive)

df_expensive
```

### Финансовая успешность: Лидеры по абсолютной прибыли

Самый дорогой фильм — далеко не всегда самый успешный. В этой витрине мы переходим от оценки затрат к реальной финансовой отдаче, вычисляя чистую разницу между мировыми кассовыми сборами и производственным бюджетом.

```python
query_profitable = "SELECT * FROM mart_most_profitable LIMIT 10"
df_profitable = query_sql(query_profitable)

df_profitable
```

### Эпический размах: Самые продолжительные картины

От финансовых показателей мы переходим к анализу формата. В этой витрине мы исследуем хронометраж фильмов. В эпоху, когда стандарты кинопроката диктуют длительность около двух часов для максимизации количества сеансов, сверхдлинные фильмы представляют особый интерес.

```python
query_longest = "SELECT * FROM mart_longest_movies LIMIT 10"
df_longest = query_sql(query_longest)

df_longest
```

### Плотность бюджета: Стоимость одной минуты фильма

В этой витрине мы не просто берем готовые цифры, а создаем новую, производную метрику — **стоимость одной минуты экранного времени**. Это позволяет оценить «плотность» инвестиций в каждый кадр, независимо от общей продолжительности картины.

```python
query_cost_per_minute = "SELECT * FROM mart_top_cost_per_minute LIMIT 10"
df_cost_per_minute = query_sql(query_cost_per_minute)

df_cost_per_minute
```

### Эффективность инвестиций: Лидеры по показателю ROI

Если абсолютная прибыль показывает общий масштаб заработка, то **ROI (Return on Investment)** демонстрирует истинную финансовую эффективность проекта. Эта метрика отвечает на главный вопрос любого продюсера или инвестора: *какой процент чистой прибыли принес каждый вложенный в производство доллар?*

```python
query_highest_roi = "SELECT * FROM mart_highest_roi LIMIT 10"
df_highest_roi = query_sql(query_highest_roi)

df_highest_roi
```

Обратная сторона финансового успеха — грандиозные кассовые провалы. В этой витрине мы ищем фильмы, которые принесли студиям максимальные убытки в абсолютном выражении, отнимая итоговые сборы от производственного бюджета.
